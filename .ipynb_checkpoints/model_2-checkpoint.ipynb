# Wet AMD Lesion Segmentation: Model 2 (Cascaded Two-Stage U-Net)
**Dataset:** AMD-SD (Optical Coherence Tomography Image Dataset for wet AMD Lesions Segmentation)
**Objective:** Construct and train a 2-part cascaded U-Net model:
1. **Stage 1 (UNet 1):** Grayscale OCT input $\to$ Continuous binary foreground probability mask ($512 \times 512$).
2. **Stage 2 (UNet 2):** Binary mask $\to$ Multi-class segmentation mask ($6$ classes, $512 \times 512$).
3. **Deep Supervision:** Joint end-to-end optimization combining **Loss B** (Binary BCE + Dice) and **Loss A** (Multi-class CE + Dice).

```python
import os
import gc
import csv
import random
import numpy as np
import cv2
import matplotlib.pyplot as plt
from tqdm import tqdm

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader

# Set environment variables for memory management
os.environ["PYTORCH_ENABLE_MPS_FALLBACK"] = "1"
os.environ["PYTORCH_MPS_HIGH_WATERMARK_RATIO"] = "0.0"

def seed_everything(seed: int = 42):
    random.seed(seed)
    os.environ["PYTHONHASHSEED"] = str(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed(seed)
        torch.cuda.manual_seed_all(seed)
        torch.backends.cudnn.deterministic = True
        torch.backends.cudnn.benchmark = False

def cleanup_memory():
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()
    elif torch.backends.mps.is_available():
        torch.mps.empty_cache()

seed_everything(42)
cleanup_memory()

device = torch.device("mps" if torch.backends.mps.is_available() else "cuda" if torch.cuda.is_available() else "cpu")
print(f"Active Compute Device: {device}")
```

```python

import os
import matplotlib as mpl
import matplotlib.pyplot as plt

FIG_WIDTH_PX = 1600                
FIG_DPI      = 100                  
FIG_DIR      = "figures"            
SAVE_FORMATS = ("png", "pdf")       
os.makedirs(FIG_DIR, exist_ok=True)

mpl.rcParams.update({
    "figure.dpi": FIG_DPI,
    "savefig.dpi": FIG_DPI,
    "pdf.fonttype": 42,            
    "ps.fonttype": 42,
    "svg.fonttype": "none",        
})

def save_fig(fname, fig=None, formats=None, width_px=None, **savefig_kw):
    """Rescale the figure to FIG_WIDTH_PX (keeping its aspect ratio) and save it in every format in SAVE_FORMATS."""
    fig = fig if fig is not None else plt.gcf()
    width_px = width_px or FIG_WIDTH_PX
    w_in, h_in = fig.get_size_inches()
    new_w = width_px / FIG_DPI
    fig.set_size_inches(new_w, h_in * new_w / w_in, forward=True)
    stem = os.path.splitext(os.path.basename(str(fname)))[0]
    paths = []
    for ext in (formats or SAVE_FORMATS):
        path = os.path.join(FIG_DIR, f"{stem}.{ext}")
        fig.savefig(path, format=ext, **savefig_kw)
        paths.append(path)
    w, h = fig.get_size_inches()
    print(f"[figure] {stem}: {int(round(w * FIG_DPI))} x {int(round(h * FIG_DPI))} px -> {', '.join(paths)}")
    return paths


if "_orig_savefig" not in globals():
    _orig_savefig = plt.savefig

def _savefig_multi(fname, *args, **kwargs):
    kwargs.pop("dpi", None)         
    kwargs.pop("format", None)
    return save_fig(fname, **kwargs)

plt.savefig = _savefig_multi
print(f"Figures will be {FIG_WIDTH_PX} px wide and saved to '{FIG_DIR}/' as {SAVE_FORMATS}.")
```

## 1. Dataset Pipeline & Color Palette Setup
Reuses the verified `AMDDataset` loader with $(512 \times 512)$ resolution, horizontal flip ($p=0.5$), and rotation ($\pm 10^\circ$). The multi-class ground truth ($0$ to $5$) will be converted on the fly during training into a binary mask for Stage 1 supervision.

```python
# Reference RGB color palette for the 6 target classes in AMD-SD
CLASS_PALETTE_RGB = np.array([
    [0, 0, 0],        # 0: Background
    [255, 0, 0],      # 1: Subretinal Fluid (SRF)
    [0, 255, 0],      # 2: Intraretinal Fluid (IRF)
    [0, 0, 255],      # 3: Pigment Epithelial Detachment (PED)
    [255, 255, 0],    # 4: Subretinal Hyperreflective Material (SHRM)
    [255, 0, 255]     # 5: Ellipsoid Zone (IS/OS)
], dtype=np.int32)

CLASS_NAMES = ["Background", "SRF", "IRF", "PED", "SHRM", "IS/OS"]

class AMDDataset(Dataset):
    def __init__(self, data_root, split="train", target_size=(512, 512), augment=True):
        self.inputs_dir = os.path.join(data_root, split, "inputs")
        self.outputs_dir = os.path.join(data_root, split, "outputs")
        self.target_size = target_size
        self.split = split
        self.augment = augment and (split == "train")

        self.filenames = sorted([
            f for f in os.listdir(self.inputs_dir)
            if f.endswith(('.png', '.jpg', '.jpeg')) and not f.startswith('.')
        ])

    def __len__(self):
        return len(self.filenames)

    def _rgb_to_class(self, mask_rgb):
        mask = np.ascontiguousarray(mask_rgb, dtype=np.int32)
        dists = np.sum((mask[:, :, None, :] - CLASS_PALETTE_RGB[None, None, :, :]) ** 2, axis=-1)
        return np.argmin(dists, axis=-1)

    def _apply_augmentations(self, img, mask):
        if random.random() > 0.5:
            img = cv2.flip(img, 1)
            mask = cv2.flip(mask, 1)

        angle = random.uniform(-10.0, 10.0)
        h, w = img.shape[:2]
        center = (w // 2, h // 2)
        M = cv2.getRotationMatrix2D(center, angle, 1.0)
        img = cv2.warpAffine(img, M, (w, h), flags=cv2.INTER_LINEAR, borderMode=cv2.BORDER_REFLECT_101)
        mask = cv2.warpAffine(mask, M, (w, h), flags=cv2.INTER_NEAREST, borderMode=cv2.BORDER_REFLECT_101)
        return img, mask

    def __getitem__(self, idx):
        in_fname = self.filenames[idx]
        img_path = os.path.join(self.inputs_dir, in_fname)
        out_fname = in_fname.replace("_input", "_output") if "_input" in in_fname else in_fname
        mask_path = os.path.join(self.outputs_dir, out_fname)

        img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)
        mask_bgr = cv2.imread(mask_path)
        if img is None or mask_bgr is None:
            raise ValueError(f"Failed to read image/mask pair: {img_path}")

        mask_rgb = cv2.cvtColor(mask_bgr, cv2.COLOR_BGR2RGB)

        img = cv2.resize(img, (self.target_size[1], self.target_size[0]), interpolation=cv2.INTER_LINEAR)
        mask_rgb = cv2.resize(mask_rgb, (self.target_size[1], self.target_size[0]), interpolation=cv2.INTER_NEAREST)

        mask_labels = self._rgb_to_class(mask_rgb)

        if self.augment:
            img, mask_labels = self._apply_augmentations(img, mask_labels)

        img_tensor = torch.from_numpy(np.ascontiguousarray(img)).unsqueeze(0).float() / 255.0
        mask_tensor = torch.from_numpy(np.ascontiguousarray(mask_labels)).long()

        return img_tensor, mask_tensor

# Setup DataLoaders matching the baseline setup
DATA_DIR = "processed_data"
MICRO_BATCH_SIZE = 2 if device.type == "mps" else 8
ACCUM_STEPS = 4 if device.type == "mps" else 1
VAL_BATCH_SIZE = 2 if device.type == "mps" else 8
NUM_WORKERS = 0 if device.type == "mps" else 4

train_dataset = AMDDataset(DATA_DIR, split="train", target_size=(512, 512), augment=True)
val_dataset = AMDDataset(DATA_DIR, split="val", target_size=(512, 512), augment=False)

train_loader = DataLoader(train_dataset, batch_size=MICRO_BATCH_SIZE, shuffle=True, num_workers=NUM_WORKERS, pin_memory=(device.type == "cuda"))
val_loader = DataLoader(val_dataset, batch_size=VAL_BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS, pin_memory=(device.type == "cuda"))

print(f"Loaded {len(train_dataset)} training pairs and {len(val_dataset)} validation pairs.")
```

## 2. Model 2 Architecture: Two-Stage Cascaded U-Net
- **Stage 1 (Binary U-Net):** Input: $1$ channel (OCT B-scan) $\to$ Output: $1$ channel (Binary logits).
- **Sigmoid Activation:** $\sigma(\text{logits}_{\text{bin}})$ produces a continuous probability map $[0.0, 1.0]$. Passing continuous probabilities maintains differentiability so gradients from Stage 2 backpropagate into Stage 1.
- **Stage 2 (Multi-Class U-Net):** Input: $1$ channel (or concatenated $[OCT, \text{Mask}_{\text{bin}}]$) $\to$ Output: $6$ class logits.

```python
class DoubleConv(nn.Module):
    """[Conv2d -> BatchNorm -> ReLU] * 2"""
    def __init__(self, in_channels: int, out_channels: int):
        super().__init__()
        self.block = nn.Sequential(
            nn.Conv2d(in_channels, out_channels, kernel_size=3, padding=1, bias=False),
            nn.BatchNorm2d(out_channels),
            nn.ReLU(inplace=True),
            nn.Conv2d(out_channels, out_channels, kernel_size=3, padding=1, bias=False),
            nn.BatchNorm2d(out_channels),
            nn.ReLU(inplace=True),
        )

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        return self.block(x)


class UNetCore(nn.Module):
    def __init__(self, in_channels: int, num_classes: int, base_channels: int = 64):
        super().__init__()
        c = base_channels
        self.pool = nn.MaxPool2d(kernel_size=2, stride=2)

        # Encoder
        self.inc = DoubleConv(in_channels, c)
        self.down1 = DoubleConv(c, c * 2)
        self.down2 = DoubleConv(c * 2, c * 4)
        self.down3 = DoubleConv(c * 4, c * 8)

        # Bottleneck
        self.bottleneck = DoubleConv(c * 8, c * 16)

        # Decoder
        self.up1 = nn.ConvTranspose2d(c * 16, c * 8, kernel_size=2, stride=2)
        self.conv_up1 = DoubleConv(c * 16, c * 8)

        self.up2 = nn.ConvTranspose2d(c * 8, c * 4, kernel_size=2, stride=2)
        self.conv_up2 = DoubleConv(c * 8, c * 4)

        self.up3 = nn.ConvTranspose2d(c * 4, c * 2, kernel_size=2, stride=2)
        self.conv_up3 = DoubleConv(c * 4, c * 2)

        self.up4 = nn.ConvTranspose2d(c * 2, c, kernel_size=2, stride=2)
        self.conv_up4 = DoubleConv(c * 2, c)

        self.outc = nn.Conv2d(c, num_classes, kernel_size=1)

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        x1 = self.inc(x)
        x2 = self.down1(self.pool(x1))
        x3 = self.down2(self.pool(x2))
        x4 = self.down3(self.pool(x3))

        m = self.bottleneck(self.pool(x4))

        d1 = self.conv_up1(torch.cat([x4, self.up1(m)], dim=1))
        d2 = self.conv_up2(torch.cat([x3, self.up2(d1)], dim=1))
        d3 = self.conv_up3(torch.cat([x2, self.up3(d2)], dim=1))
        d4 = self.conv_up4(torch.cat([x1, self.up4(d3)], dim=1))

        return self.outc(d4)


class CascadedUNetModel2(nn.Module):
    def __init__(self, in_channels: int = 1, num_classes: int = 6, base_channels: int = 64, concatenate_input: bool = False):
        """
        Model 2: Two-stage cascaded network.
        - unet1: grayscale OCT (1 ch) -> binary mask logits (1 ch)
        - unet2: binary mask (1 ch, or 2 ch if concatenated with grayscale) -> 6-class logits
        """
        super().__init__()
        self.concatenate_input = concatenate_input

        # Stage 1: Predict binary foreground mask (1 channel logit)
        self.unet1 = UNetCore(in_channels=in_channels, num_classes=1, base_channels=base_channels)

        # Stage 2: Predict 6-class segmentation
        stage2_in_channels = (in_channels + 1) if concatenate_input else 1
        self.unet2 = UNetCore(in_channels=stage2_in_channels, num_classes=num_classes, base_channels=base_channels)

    def forward(self, x: torch.Tensor):
        # 1. Forward through Stage 1 U-Net
        bin_logits = self.unet1(x)                                 # Shape: [B, 1, H, W]
        bin_prob = torch.sigmoid(bin_logits)                       # Continuous probability map [0, 1]

        # 2. Prepare input for Stage 2
        if self.concatenate_input:
            stage2_input = torch.cat([x, bin_prob], dim=1)         # Shape: [B, 2, H, W]
        else:
            stage2_input = bin_prob                                # Shape: [B, 1, H, W]

        # 3. Forward through Stage 2 U-Net
        multi_logits = self.unet2(stage2_input)                    # Shape: [B, 6, H, W]

        return bin_logits, multi_logits
```

## 3. Loss Functions for Cascaded Supervision
- **Loss B (Stage 1 Binary Loss):** Evaluates predicted binary logits against binarized ground-truth targets ($y_{\text{bin}} = (y_{\text{multi}} > 0)$) using Binary Cross Entropy (BCE) with Logits + Soft Binary Dice Loss:
  $$\mathcal{L}_B = 0.5 \cdot \text{BCEWithLogits} + 0.5 \cdot \text{Dice}_{\text{bin}}$$
- **Loss A (Stage 2 Multi-Class Loss):** Evaluates predicted 6-class logits against the multi-class ground-truth mask using the exact combined `CE_DiceLoss` from `main.ipynb`:
  $$\mathcal{L}_A = 0.5 \cdot \mathcal{L}_{\text{CE}} + 0.5 \cdot \mathcal{L}_{\text{Dice}}$$
- **Total Combined Loss:**
  $$\mathcal{L}_{\text{Total}} = \mathcal{L}_A + \lambda_B \cdot \mathcal{L}_B \quad (\lambda_B = 1.0)$$

```python
class BinaryDiceLoss(nn.Module):
    def __init__(self, smooth: float = 1.0):
        super().__init__()
        self.smooth = smooth

    def forward(self, probs: torch.Tensor, targets: torch.Tensor) -> torch.Tensor:
        probs = probs.view(-1)
        targets = targets.view(-1)
        intersection = (probs * targets).sum()
        dice = (2.0 * intersection + self.smooth) / (probs.sum() + targets.sum() + self.smooth)
        return 1.0 - dice


class BinaryBCEDiceLoss(nn.Module):
    def __init__(self, bce_weight: float = 0.5, dice_weight: float = 0.5):
        super().__init__()
        self.bce_weight = bce_weight
        self.dice_weight = dice_weight
        self.bce = nn.BCEWithLogitsLoss()
        self.dice = BinaryDiceLoss()

    def forward(self, logits: torch.Tensor, targets: torch.Tensor) -> torch.Tensor:
        loss_bce = self.bce(logits, targets)
        probs = torch.sigmoid(logits)
        loss_dice = self.dice(probs, targets)
        return (self.bce_weight * loss_bce) + (self.dice_weight * loss_dice)


class CE_DiceLoss(nn.Module):
    def __init__(self, num_classes: int = 6, ignore_index: int = 255, ce_weight: float = 0.5, dice_weight: float = 0.5, smooth: float = 1.0):
        super().__init__()
        self.num_classes = num_classes
        self.ignore_index = ignore_index
        self.ce_weight = ce_weight
        self.dice_weight = dice_weight
        self.smooth = smooth
        self.ce = nn.CrossEntropyLoss(ignore_index=ignore_index)

    def forward(self, logits: torch.Tensor, targets: torch.Tensor) -> torch.Tensor:
        ce_loss = self.ce(logits, targets)

        probs = F.softmax(logits, dim=1)
        valid_mask = (targets != self.ignore_index)
        targets_clean = targets.clone()
        targets_clean[~valid_mask] = 0

        targets_one_hot = F.one_hot(targets_clean, num_classes=self.num_classes).permute(0, 3, 1, 2).float()

        dice_total = 0.0
        for c in range(1, self.num_classes):
            p = probs[:, c, ...][valid_mask]
            t = targets_one_hot[:, c, ...][valid_mask]
            intersection = (p * t).sum()
            dice = (2.0 * intersection + self.smooth) / (p.sum() + t.sum() + self.smooth)
            dice_total += (1.0 - dice)

        dice_loss = dice_total / (self.num_classes - 1)
        return (self.ce_weight * ce_loss) + (self.dice_weight * dice_loss)


class CascadedLossModel2(nn.Module):
    def __init__(self, lambda_bin: float = 1.0, ce_weight: float = 0.5, dice_weight: float = 0.5):
        super().__init__()
        self.lambda_bin = lambda_bin
        self.loss_b_criterion = BinaryBCEDiceLoss(bce_weight=ce_weight, dice_weight=dice_weight)
        self.loss_a_criterion = CE_DiceLoss(num_classes=6, ignore_index=255, ce_weight=ce_weight, dice_weight=dice_weight)

    def forward(self, bin_logits: torch.Tensor, multi_logits: torch.Tensor, targets_multi: torch.Tensor):
        # Convert multi-class targets (0 to 5) to binary foreground target (0 or 1)
        targets_bin = (targets_multi > 0).float().unsqueeze(1)    # Shape: [B, 1, H, W]

        loss_b = self.loss_b_criterion(bin_logits, targets_bin)
        loss_a = self.loss_a_criterion(multi_logits, targets_multi)

        total_loss = loss_a + (self.lambda_bin * loss_b)
        return total_loss, loss_a, loss_b
```

## 4. Evaluation Metrics & Learning Rate Scheduler
Includes `GlobalMetricTracker` to accumulate intersection and union counts across all validation scans for the 5 target lesion classes, and `PolyLRScheduler` matching `main.ipynb`.

```python
class PolyLRScheduler(torch.optim.lr_scheduler._LRScheduler):
    def __init__(self, optimizer, max_epochs, power=0.9, min_lr=1e-6, last_epoch=-1):
        self.max_epochs = max_epochs
        self.power = power
        self.min_lr = min_lr
        super().__init__(optimizer, last_epoch)

    def get_lr(self):
        factor = (1.0 - self.last_epoch / max(1, self.max_epochs)) ** self.power
        return [max(self.min_lr, base_lr * factor) for base_lr in self.base_lrs]


class GlobalMetricTracker:
    def __init__(self, num_classes: int = 6):
        self.num_classes = num_classes
        self.class_map = {1: "SRF", 2: "IRF", 3: "PED", 4: "SHRM", 5: "IS/OS"}
        self.reset()

    def reset(self):
        self.intersections = {c: 0.0 for c in range(1, self.num_classes)}
        self.total_preds = {c: 0.0 for c in range(1, self.num_classes)}
        self.total_targets = {c: 0.0 for c in range(1, self.num_classes)}

    def update(self, preds: torch.Tensor, targets: torch.Tensor):
        for c in range(1, self.num_classes):
            p = (preds == c).float()
            t = (targets == c).float()
            self.intersections[c] += (p * t).sum().item()
            self.total_preds[c] += p.sum().item()
            self.total_targets[c] += t.sum().item()

    def compute(self):
        dice_scores = {}
        iou_scores = {}
        for c, name in self.class_map.items():
            inter = self.intersections[c]
            p_sum = self.total_preds[c]
            t_sum = self.total_targets[c]
            union = p_sum + t_sum - inter

            dice = (2.0 * inter) / (p_sum + t_sum + 1e-8) * 100.0
            iou = (inter) / (union + 1e-8) * 100.0
            dice_scores[name] = dice
            iou_scores[name] = iou

        mean_dice = sum(dice_scores.values()) / len(dice_scores)
        mean_iou = sum(iou_scores.values()) / len(iou_scores)
        return dice_scores, iou_scores, mean_dice, mean_iou
```

## 5. Model 2 Training & Checkpointing Engine
Runs the end-to-end training loop across 20 epochs using AdamW ($\eta = 1 \times 10^{-4}$, weight decay $= 1 \times 10^{-4}$), gradient accumulation, and saves the optimal weights to `best_unet_model2.pth`.

```python
def run_model2_training():
    cleanup_memory()
    seed_everything(42)

    EPOCHS = 35
    LEARNING_RATE = 1e-4
    NUM_CLASSES = 6
    LOG_FILE = "training_log_model2.csv"
    CKPT_PATH = "best_unet_model2.pth"

    # 1. Instantiate Model 2, Criterion, Optimizer, Scheduler
    model = CascadedUNetModel2(
        in_channels=1,
        num_classes=NUM_CLASSES,
        base_channels=64,
        concatenate_input=False  # Set True if you want UNet 2 to receive [OCT, BinaryMask]
    ).to(device)

    cascade_criterion = CascadedLossModel2(lambda_bin=1.0, ce_weight=0.5, dice_weight=0.5).to(device)
    optimizer = torch.optim.AdamW(model.parameters(), lr=LEARNING_RATE, weight_decay=1e-4)
    scheduler = PolyLRScheduler(optimizer, max_epochs=EPOCHS, power=0.9, min_lr=1e-6)

    # 2. Setup CSV Logger
    fieldnames = [
        "epoch", "train_total_loss", "train_loss_a", "train_loss_b",
        "val_total_loss", "val_loss_a", "val_loss_b", "mean_dice", "mean_iou",
        "SRF_dice", "IRF_dice", "PED_dice", "SHRM_dice", "IS_OS_dice",
        "SRF_iou", "IRF_iou", "PED_iou", "SHRM_iou", "IS_OS_iou"
    ]
    with open(LOG_FILE, mode="w", newline="") as f:
        writer = csv.DictWriter(f, fieldnames=fieldnames)
        writer.writeheader()

    best_mean_dice = 0.0

    print("Beginning Model 2 (Cascaded U-Net) Training...")
    for epoch in range(1, EPOCHS + 1):
        model.train()
        train_tot_accum = 0.0
        train_a_accum = 0.0
        train_b_accum = 0.0
        optimizer.zero_grad(set_to_none=True)

        pbar = tqdm(train_loader, desc=f"Epoch {epoch:02d}/{EPOCHS} [Train]")
        for step, (images, targets) in enumerate(pbar):
            images = images.to(device)
            targets = targets.to(device, dtype=torch.long)

            bin_logits, multi_logits = model(images)
            total_loss, loss_a, loss_b = cascade_criterion(bin_logits, multi_logits, targets)

            # Gradient accumulation scaling
            scaled_loss = total_loss / ACCUM_STEPS
            scaled_loss.backward()

            if (step + 1) % ACCUM_STEPS == 0 or (step + 1) == len(train_loader):
                optimizer.step()
                optimizer.zero_grad(set_to_none=True)

            train_tot_accum += total_loss.item()
            train_a_accum += loss_a.item()
            train_b_accum += loss_b.item()
            pbar.set_postfix({"tot_loss": f"{total_loss.item():.4f}", "L_a": f"{loss_a.item():.3f}", "L_b": f"{loss_b.item():.3f}"})

        scheduler.step()
        epoch_train_tot = train_tot_accum / len(train_loader)
        epoch_train_a = train_a_accum / len(train_loader)
        epoch_train_b = train_b_accum / len(train_loader)

        # 3. Validation Phase
        model.eval()
        val_tot_accum = 0.0
        val_a_accum = 0.0
        val_b_accum = 0.0
        tracker = GlobalMetricTracker(num_classes=NUM_CLASSES)

        with torch.no_grad():
            for images, targets in tqdm(val_loader, desc=f"Epoch {epoch:02d}/{EPOCHS} [Val]"):
                images = images.to(device)
                targets = targets.to(device, dtype=torch.long)

                bin_logits, multi_logits = model(images)
                total_loss, loss_a, loss_b = cascade_criterion(bin_logits, multi_logits, targets)

                val_tot_accum += total_loss.item()
                val_a_accum += loss_a.item()
                val_b_accum += loss_b.item()

                preds = torch.argmax(multi_logits, dim=1)
                tracker.update(preds, targets)

        epoch_val_tot = val_tot_accum / len(val_loader)
        epoch_val_a = val_a_accum / len(val_loader)
        epoch_val_b = val_b_accum / len(val_loader)
        avg_dice, avg_iou, mean_dice, mean_iou = tracker.compute()

        print(f"\n--- Epoch {epoch:02d} Summary (Model 2) ---")
        print(f"Train Total Loss: {epoch_train_tot:.4f} (L_a: {epoch_train_a:.4f}, L_b: {epoch_train_b:.4f})")
        print(f"Val Total Loss:   {epoch_val_tot:.4f} (L_a: {epoch_val_a:.4f}, L_b: {epoch_val_b:.4f})")
        print(f"Validation Mean Dice: {mean_dice:.2f}% | Mean IoU: {mean_iou:.2f}%")
        print(f"Per-Class Dice -> SRF: {avg_dice['SRF']:.2f}% | IRF: {avg_dice['IRF']:.2f}% | "
              f"PED: {avg_dice['PED']:.2f}% | SHRM: {avg_dice['SHRM']:.2f}% | IS/OS: {avg_dice['IS/OS']:.2f}%")

        # 4. Log to CSV
        log_entry = {
            "epoch": epoch,
            "train_total_loss": f"{epoch_train_tot:.4f}",
            "train_loss_a": f"{epoch_train_a:.4f}",
            "train_loss_b": f"{epoch_train_b:.4f}",
            "val_total_loss": f"{epoch_val_tot:.4f}",
            "val_loss_a": f"{epoch_val_a:.4f}",
            "val_loss_b": f"{epoch_val_b:.4f}",
            "mean_dice": f"{mean_dice:.2f}",
            "mean_iou": f"{mean_iou:.2f}",
            "SRF_dice": f"{avg_dice['SRF']:.2f}",
            "IRF_dice": f"{avg_dice['IRF']:.2f}",
            "PED_dice": f"{avg_dice['PED']:.2f}",
            "SHRM_dice": f"{avg_dice['SHRM']:.2f}",
            "IS_OS_dice": f"{avg_dice['IS/OS']:.2f}",
            "SRF_iou": f"{avg_iou['SRF']:.2f}",
            "IRF_iou": f"{avg_iou['IRF']:.2f}",
            "PED_iou": f"{avg_iou['PED']:.2f}",
            "SHRM_iou": f"{avg_iou['SHRM']:.2f}",
            "IS_OS_iou": f"{avg_iou['IS/OS']:.2f}"
        }
        with open(LOG_FILE, mode="a", newline="") as f:
            writer = csv.DictWriter(f, fieldnames=fieldnames)
            writer.writerow(log_entry)

        # 5. Checkpointing
        if mean_dice > best_mean_dice:
            best_mean_dice = mean_dice
            torch.save(model.state_dict(), CKPT_PATH)
            print(f"--> Saved optimal Model 2 checkpoint with Mean Dice: {best_mean_dice:.2f}%")

        cleanup_memory()

    print(f"\nModel 2 Training complete. Peak Validation Mean Dice: {best_mean_dice:.2f}%")

# Execute training
run_model2_training()
```

## 8. Statistical Evaluation & Per-Class Biomarker Distribution Analysis

Evaluates the saved model checkpoint on individual OCT B-scans to analyze variance, class-level difficulty, and statistical robustness:

1. **Per-Slice Metric Computation:** Calculates slice-level Dice and IoU scores across all 5 target biomarker classes (`SRF`, `IRF`, `PED`, `SHRM`, `IS/OS`). Slices where a biomarker is absent in both ground truth and prediction are assigned `NaN` to prevent skewing the evaluation.
2. **Empirical Bootstrapped 99% Confidence Intervals:** Runs non-parametric resampling with replacement ($N = 2{,}000$ iterations) to construct 99% confidence intervals ($\alpha = 0.01$) for mean Dice and IoU without assuming metric normality.
3. **Distribution Visualizations (Boxplots):** 
   - **Overall Slice Variance:** Boxplots showing the spread and mean (diamond marker) of per-image Dice and IoU.
   - **Biomarker Breakdown:** Class-stratified boxplots highlighting which anatomical lesions benefit most from the architectural design choices.

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import torch
from torch.utils.data import DataLoader
from tqdm import tqdm

# --- 1. Load Trained Checkpoint ---
CKPT_PATH = "best_unet_model2.pth"

model_2 = CascadedUNetModel2(
    in_channels=1,
    num_classes=6,
    base_channels=64,
    concatenate_input=False
).to(device)

checkpoint = torch.load(CKPT_PATH, map_location=device)
model_2.load_state_dict(checkpoint)
model_2.eval()
print(f"Loaded weights from {CKPT_PATH} into Model 2 successfully.")

# Setup test/eval loader
if "test_loader" in globals():
    eval_loader = test_loader
    print("Evaluating on test_loader.")
elif os.path.exists(os.path.join(DATA_DIR, "test")):
    test_dataset = AMDDataset(DATA_DIR, split="test", target_size=(512, 512), augment=False)
    eval_loader = DataLoader(
        test_dataset,
        batch_size=VAL_BATCH_SIZE,
        shuffle=False,
        num_workers=NUM_WORKERS,
        pin_memory=(device.type == "cuda")
    )
    print(f"Loaded {len(test_dataset)} test pairs.")
else:
    eval_loader = val_loader
    print(f"Evaluating on {len(val_dataset)} validation pairs.")


# --- 2. Metric Calculation Helpers ---
def compute_image_metrics(pred_mask: np.ndarray, target_mask: np.ndarray):
    class_map = {1: "SRF", 2: "IRF", 3: "PED", 4: "SHRM", 5: "IS/OS"}
    dice_scores = {}
    iou_scores = {}

    for c_idx, c_name in class_map.items():
        p = (pred_mask == c_idx)
        t = (target_mask == c_idx)
        p_sum, t_sum = p.sum(), t.sum()
        intersection = np.logical_and(p, t).sum()
        union = np.logical_or(p, t).sum()

        if t_sum == 0 and p_sum == 0:
            dice_scores[c_name] = np.nan
            iou_scores[c_name] = np.nan
        else:
            dice = (2.0 * intersection) / (p_sum + t_sum + 1e-8) * 100.0
            iou = intersection / (union + 1e-8) * 100.0
            dice_scores[c_name] = dice
            iou_scores[c_name] = iou

    valid_dices = [v for v in dice_scores.values() if not np.isnan(v)]
    valid_ious = [v for v in iou_scores.values() if not np.isnan(v)]

    mean_dice = np.mean(valid_dices) if len(valid_dices) > 0 else np.nan
    mean_iou = np.mean(valid_ious) if len(valid_ious) > 0 else np.nan

    return dice_scores, iou_scores, mean_dice, mean_iou


def bootstrap_ci(data: np.ndarray, n_bootstraps: int = 2000, ci: float = 0.99):
    clean_data = data[~np.isnan(data)]
    if len(clean_data) == 0:
        return np.nan, (np.nan, np.nan)

    n = len(clean_data)
    boot_means = np.empty(n_bootstraps)
    for i in range(n_bootstraps):
        sample = np.random.choice(clean_data, size=n, replace=True)
        boot_means[i] = np.mean(sample)

    alpha = (1.0 - ci) / 2.0
    ci_lower = np.percentile(boot_means, alpha * 100.0)
    ci_upper = np.percentile(boot_means, (1.0 - alpha) * 100.0)
    return float(np.mean(clean_data)), (float(ci_lower), float(ci_upper))


# --- 3. Evaluate Models ---
models_to_evaluate = {
    "Model 2 (Cascaded U-Net)": model_2
}

per_image_records = []
lesions = ["SRF", "IRF", "PED", "SHRM", "IS/OS"]

for model_name, net in models_to_evaluate.items():
    net.eval()
    img_counter = 0

    with torch.no_grad():
        for images, targets in tqdm(eval_loader, desc=f"Evaluating {model_name}"):
            images = images.to(device)
            targets = targets.cpu().numpy()

            outputs = net(images)
            multi_logits = outputs[1] if isinstance(outputs, tuple) else outputs
            preds = torch.argmax(multi_logits, dim=1).cpu().numpy()

            batch_size = images.size(0)
            for b in range(batch_size):
                p_img = preds[b]
                t_img = targets[b]

                d_scores, i_scores, m_dice, m_iou = compute_image_metrics(p_img, t_img)

                record = {
                    "Model": model_name,
                    "Image_ID": img_counter,
                    "Mean_Dice": m_dice,
                    "Mean_IoU": m_iou,
                }
                for l in lesions:
                    record[f"{l}_Dice"] = d_scores[l]
                    record[f"{l}_IoU"] = i_scores[l]

                per_image_records.append(record)
                img_counter += 1

df_results = pd.DataFrame(per_image_records)


# --- 4. 99% Bootstrapped Confidence Interval Summary ---
summary_rows = []
for model_name in models_to_evaluate.keys():
    sub_df = df_results[df_results["Model"] == model_name]

    m_dice, (d_low, d_high) = bootstrap_ci(sub_df["Mean_Dice"].values, n_bootstraps=2000, ci=0.99)
    m_iou, (i_low, i_high) = bootstrap_ci(sub_df["Mean_IoU"].values, n_bootstraps=2000, ci=0.99)

    row = {
        "Model": model_name,
        "Mean Dice [%] (99% CI)": f"{m_dice:.2f} [{d_low:.2f}, {d_high:.2f}]",
        "Mean IoU [%] (99% CI)": f"{m_iou:.2f} [{i_low:.2f}, {i_high:.2f}]",
    }

    for l in lesions:
        l_dice, (ld_low, ld_high) = bootstrap_ci(sub_df[f"{l}_Dice"].values, n_bootstraps=2000, ci=0.99)
        l_iou, (li_low, li_high) = bootstrap_ci(sub_df[f"{l}_IoU"].values, n_bootstraps=2000, ci=0.99)
        row[f"{l} Dice [%]"] = f"{l_dice:.2f} [{ld_low:.2f}, {ld_high:.2f}]"
        row[f"{l} IoU [%]"] = f"{l_iou:.2f} [{li_low:.2f}, {li_high:.2f}]"

    summary_rows.append(row)

df_bootstrap_summary = pd.DataFrame(summary_rows)
display(df_bootstrap_summary.T)


# --- 5. Boxplots ---
sns.set_theme(style="whitegrid", palette="Set2")
plt.rcParams.update({"font.sans-serif": "DejaVu Sans", "font.size": 11})

# Overall Mean Metrics
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.boxplot(
    data=df_results, x="Model", y="Mean_Dice", ax=axes[0],
    showmeans=True, meanprops={"marker": "D", "markerfacecolor": "red", "markeredgecolor": "red"}
)
axes[0].set_title("Per-Image Mean Dice Distribution (Red Diamond = Mean)", fontweight="bold")
axes[0].set_ylabel("Dice Score [%]")
axes[0].set_ylim(-5, 105)

sns.boxplot(
    data=df_results, x="Model", y="Mean_IoU", ax=axes[1],
    showmeans=True, meanprops={"marker": "D", "markerfacecolor": "red", "markeredgecolor": "red"}
)
axes[1].set_title("Per-Image Mean IoU Distribution (Red Diamond = Mean)", fontweight="bold")
axes[1].set_ylabel("IoU Score [%]")
axes[1].set_ylim(-5, 105)

plt.tight_layout()
plt.show()

# Per-Class Lesion Breakdown Across Slices
dice_cols = [f"{l}_Dice" for l in lesions]
df_melted_dice = pd.melt(
    df_results,
    id_vars=["Model", "Image_ID"],
    value_vars=dice_cols,
    var_name="Lesion_Class",
    value_name="Dice_Score"
).dropna()

df_melted_dice["Lesion_Class"] = df_melted_dice["Lesion_Class"].apply(lambda name: name.replace("_Dice", ""))

plt.figure(figsize=(15, 6))
sns.boxplot(
    data=df_melted_dice,
    x="Lesion_Class",
    y="Dice_Score",
    hue="Model",
    showmeans=True,
    meanprops={"marker": "o", "markerfacecolor": "darkred", "markeredgecolor": "black", "markersize": 5}
)
plt.title("Per-Class Lesion Segmentation Performance Distribution Across Slices", fontsize=14, fontweight="bold")
plt.xlabel("Biomarker Class", fontsize=12)
plt.ylabel("Dice Score [%]", fontsize=12)
plt.ylim(-5, 105)
plt.legend(title="Architecture", loc="lower right", frameon=True)
plt.tight_layout()
plt.show()
```

## 9. Top-2 Predictions (Final Analysis)
Reports how much better Model 2 looks when the **two** most probable classes of the Stage 2 (multi-class) output are allowed: a pixel counts as correct if the ground-truth class is the top-1 *or* the 2nd most probable class. If neither is correct, the top-1 label is kept as the prediction.

The cell prints:
* top-1 vs top-2 pixel-pooled Dice/IoU per lesion class, the unweighted mean of the 5 lesion classes, and the share of missed pixels that the 2nd choice recovers,
* 99% bootstrap confidence intervals for the per-image top-2 scores,
* a top-1 vs top-2 bar chart (bar colour = lesion class; top-2 hatched),
* the slices where the 2nd choice rescues the most lesion pixels (Input | Ground truth | Top-1 | 2nd most probable class | Rescued pixels).

The evaluation uses the `test` split if `processed_data/test/inputs` exists, otherwise `val` (with a warning). This analysis uses the ground truth, so treat it as a measure of how close the model is, not as a deployable result. All tables are saved to `eval_outputs/`.

```python
# ======================= Top-2 predictions (final analysis) =======================
# Top-2 rule: a pixel counts as correct if the ground-truth class is one of the model's TWO most probable classes.
# This uses the ground truth, so it measures how close the model is, not a deployable result.
import os
import numpy as np
import pandas as pd
import torch
import matplotlib.pyplot as plt
from matplotlib.patches import Patch
from torch.utils.data import DataLoader
from tqdm.auto import tqdm

T2_DATA_DIR = DATA_DIR
T2_CKPT     = "best_unet_model2.pth"
T2_KEY      = "model2_cascaded"
T2_LABEL    = "Model 2 (Cascaded U-Net)"
T2_OUT_DIR  = "eval_outputs"
T2_NC       = 6
T2_LESIONS  = CLASS_NAMES[1:]
T2_COLORS   = {n: "#%02x%02x%02x" % tuple(int(v) for v in rgb) for n, rgb in zip(CLASS_NAMES, CLASS_PALETTE_RGB)}
T2_COLORS["Mean"] = "#333333"
T2_PALETTE  = np.array(CLASS_PALETTE_RGB, dtype=np.uint8)
T2_ORANGE   = (255, 165, 0)
os.makedirs(T2_OUT_DIR, exist_ok=True)

# ---------------- evaluation split: test if available, otherwise val ----------------
if os.path.exists(os.path.join(T2_DATA_DIR, "test", "inputs")):
    T2_SPLIT = "test"
else:
    T2_SPLIT = "val"
    print("WARNING: 'processed_data/test/inputs' not found -> using the 'val' split "
          "(it was used for checkpoint selection, so numbers are slightly optimistic).")


# ---------------- one inference pass -> per-image confusion matrices (top-1 and top-2) ----------------
@torch.no_grad()
def t2_collect(model, loader):
    cms1, cms2 = [], []
    for images, targets in tqdm(loader, desc=f"Evaluating {T2_KEY} [{T2_SPLIT}]"):
        _, multi_logits = model(images.to(device))                    # Stage 2 (6-class) output
        top2 = multi_logits.topk(2, dim=1).indices.cpu()              # (B, 2, H, W)
        pred1 = top2[:, 0]
        hit2 = (top2 == targets.unsqueeze(1)).any(dim=1)              # GT is the 1st or 2nd most probable class
        pred2 = torch.where(hit2, targets, pred1)                     # otherwise keep the top-1 label
        for b in range(pred1.shape[0]):
            t = targets[b].reshape(-1)
            cms1.append(torch.bincount(t * T2_NC + pred1[b].reshape(-1), minlength=T2_NC ** 2).reshape(T2_NC, T2_NC).numpy())
            cms2.append(torch.bincount(t * T2_NC + pred2[b].reshape(-1), minlength=T2_NC ** 2).reshape(T2_NC, T2_NC).numpy())
    return np.stack(cms1).astype(np.int64), np.stack(cms2).astype(np.int64)


t2_ds = AMDDataset(T2_DATA_DIR, split=T2_SPLIT, target_size=(512, 512), augment=False)
t2_loader = DataLoader(t2_ds, batch_size=VAL_BATCH_SIZE, shuffle=False, num_workers=0)
t2_net = CascadedUNetModel2(in_channels=1, num_classes=T2_NC, base_channels=64, concatenate_input=False).to(device)
t2_net.load_state_dict(torch.load(T2_CKPT, map_location=device))
t2_net.eval()
cms1, cms2 = t2_collect(t2_net, t2_loader)
filenames = np.array(t2_ds.filenames)


# ---------------- metrics from confusion matrices ----------------
def t2_pooled_table(cms):
    tot = cms.sum(axis=0).astype(np.float64)
    rows = []
    for c, name in enumerate(T2_LESIONS, start=1):
        inter, gt, pr = tot[c, c], tot[c, :].sum(), tot[:, c].sum()
        rows.append({"Class": name, "GT px": int(gt),
                     "Dice (%)": 200.0 * inter / max(gt + pr, 1.0),
                     "IoU (%)": 100.0 * inter / max(gt + pr - inter, 1.0)})
    df = pd.DataFrame(rows)
    df["GT share of lesion px (%)"] = 100.0 * df["GT px"] / df["GT px"].sum()
    return df


def t2_per_image(cms):
    """Per-image Dice / IoU (%), convention 'exclude': class absent in both GT and prediction -> NaN."""
    out = {}
    for c, name in enumerate(T2_LESIONS, start=1):
        inter = cms[:, c, c].astype(np.float64)
        t_sum = cms[:, c, :].sum(axis=1).astype(np.float64)
        p_sum = cms[:, :, c].sum(axis=1).astype(np.float64)
        denom, union = t_sum + p_sum, t_sum + p_sum - inter
        out[f"{name}_DSC"] = np.where(denom > 0, 2.0 * inter / np.maximum(denom, 1.0), np.nan) * 100.0
        out[f"{name}_IoU"] = np.where(union > 0, inter / np.maximum(union, 1.0), np.nan) * 100.0
    df = pd.DataFrame(out)
    df["Mean_DSC"] = df[[f"{n}_DSC" for n in T2_LESIONS]].mean(axis=1, skipna=True)
    df["Mean_IoU"] = df[[f"{n}_IoU" for n in T2_LESIONS]].mean(axis=1, skipna=True)
    return df


def t2_bootstrap_ci(values, n_boot=2000, ci=0.99, seed=42, chunk=250):
    v = np.asarray(values, dtype=np.float64); v = v[~np.isnan(v)]
    if v.size == 0:
        return np.nan, np.nan, np.nan, 0
    rng = np.random.default_rng(seed); means = np.empty(n_boot)
    for s in range(0, n_boot, chunk):
        e = min(n_boot, s + chunk)
        means[s:e] = v[rng.integers(0, v.size, size=(e - s, v.size))].mean(axis=1)
    a = (1.0 - ci) / 2.0 * 100.0
    return float(v.mean()), float(np.percentile(means, a)), float(np.percentile(means, 100.0 - a)), int(v.size)


def t2_overlay(img, mask, alpha=0.45, binary=False):
    """Semi-transparent mask on the grayscale OCT slice. img: float [0,1] (H,W); mask: int labels (H,W)."""
    base = np.repeat((np.clip(img, 0, 1) * 255.0)[..., None], 3, axis=-1).astype(np.float32)
    m = np.asarray(mask) > 0
    color = np.zeros_like(base)
    if binary:
        color[:] = np.array(T2_ORANGE, dtype=np.float32)
    else:
        color = T2_PALETTE[np.asarray(mask).astype(int)].astype(np.float32)
    out = base.copy()
    out[m] = (1.0 - alpha) * base[m] + alpha * color[m]
    return out.astype(np.uint8)


def t2_save(stem, fig):
    if "save_fig" in globals():
        save_fig(stem, fig=fig)
    else:
        fig.savefig(f"{stem}.png", dpi=200, bbox_inches="tight")


# ---------------- 1. top-1 vs top-2 table ----------------
t1, t2 = t2_pooled_table(cms1), t2_pooled_table(cms2)
tot1, tot2 = cms1.sum(axis=0).astype(np.float64), cms2.sum(axis=0).astype(np.float64)
cats = T2_LESIONS + ["Mean"]

d1, d2 = list(t1["Dice (%)"]) + [t1["Dice (%)"].mean()], list(t2["Dice (%)"]) + [t2["Dice (%)"].mean()]
i1, i2 = list(t1["IoU (%)"]) + [t1["IoU (%)"].mean()], list(t2["IoU (%)"]) + [t2["IoU (%)"].mean()]
recovered = []                      # share of GT pixels missed by top-1 that the 2nd choice gets right
for c in range(1, T2_NC):
    missed = tot1[c, :].sum() - tot1[c, c]
    recovered.append(100.0 * (tot2[c, c] - tot1[c, c]) / max(missed, 1.0))
recovered.append(float(np.mean(recovered)))

top2_df = pd.DataFrame({"Class": cats, "Dice top-1 (%)": d1, "Dice top-2 (%)": d2, "Dice gain (pts)": np.array(d2) - np.array(d1),
                        "IoU top-1 (%)": i1, "IoU top-2 (%)": i2, "IoU gain (pts)": np.array(i2) - np.array(i1),
                        "Missed px recovered by 2nd choice (%)": recovered}).round(2)
print(f"{T2_LABEL} | split='{T2_SPLIT}' | {len(cms1)} slices | top-1 vs top-2 (pixel-pooled; Mean = unweighted mean of the 5 lesion classes)")
print(top2_df.to_string(index=False))
top2_df.to_csv(os.path.join(T2_OUT_DIR, f"{T2_KEY}_{T2_SPLIT}_top1_vs_top2.csv"), index=False)

# ---------------- 2. 99% bootstrap CIs of the per-image top-2 scores ----------------
scores_top2 = pd.concat([pd.DataFrame({"filename": filenames}), t2_per_image(cms2)], axis=1)
scores_top2.to_csv(os.path.join(T2_OUT_DIR, f"{T2_KEY}_{T2_SPLIT}_scores_top2_exclude.csv"), index=False)
ci_cols = [f"{n}_DSC" for n in T2_LESIONS] + ["Mean_DSC"] + [f"{n}_IoU" for n in T2_LESIONS] + ["Mean_IoU"]
ci_rows = []
for col in ci_cols:
    m, lo, hi, n = t2_bootstrap_ci(scores_top2[col].to_numpy())
    ci_rows.append({"Column": col, "N": n, "Mean": m, "CI lower": lo, "CI upper": hi})
ci_top2 = pd.DataFrame(ci_rows).round(2)
ci_top2.to_csv(os.path.join(T2_OUT_DIR, f"{T2_KEY}_{T2_SPLIT}_ci99_top2.csv"), index=False)
print("\n99% bootstrap CIs of the per-image mean, top-2 predictions (convention='exclude'):")
print(ci_top2.to_string(index=False))

# ---------------- 3. bar chart: top-1 (solid) vs top-2 (hatched), bar colour = lesion class ----------------
fig, axes = plt.subplots(1, 2, figsize=(16, 5.5))
x, w = np.arange(len(cats)), 0.38
for ax, metric, a, b in zip(axes, ["Dice (DSC)", "IoU"], [d1, i1], [d2, i2]):
    cols = [T2_COLORS[c] for c in cats]
    ax.bar(x - w / 2, a, w, color=cols, edgecolor="black")
    r2 = ax.bar(x + w / 2, b, w, color=cols, edgecolor="black", hatch="//", alpha=0.8)
    for bar in r2:
        ax.annotate(f"{bar.get_height():.1f}", (bar.get_x() + bar.get_width() / 2, bar.get_height()), xytext=(0, 3),
                    textcoords="offset points", ha="center", va="bottom", fontsize=8, fontweight="bold")
    ax.set_xticks(x); ax.set_xticklabels(cats); ax.set_ylim(0, 105)
    ax.set_ylabel(f"{metric} (%)"); ax.grid(axis="y", linestyle="--", alpha=0.5)
    ax.set_title(f"{T2_LABEL}: {metric}, top-1 vs top-2 ({T2_SPLIT})", fontweight="bold")
axes[0].legend(handles=[Patch(facecolor="white", edgecolor="black", label="Top-1 (solid)"),
                        Patch(facecolor="white", edgecolor="black", hatch="//", label="Top-2 (hatched)")], loc="lower right")
plt.tight_layout()
t2_save(f"{T2_KEY}_{T2_SPLIT}_top1_vs_top2", fig)
plt.show()

# ---------------- 4. slices where the 2nd choice rescues the most lesion pixels ----------------
diag = np.arange(1, T2_NC)
rescued = (cms2[:, diag, diag] - cms1[:, diag, diag]).sum(axis=1)
idx_list = [int(i) for i in np.argsort(-rescued)[:3] if rescued[i] > 0]
if idx_list:
    heads = ["Input OCT", "Ground truth", "Top-1 prediction", "2nd most probable class", "Rescued by 2nd choice (orange)"]
    fig, axs = plt.subplots(len(idx_list), 5, figsize=(16, 3.3 * len(idx_list) + 0.9), squeeze=False)
    for r, i in enumerate(idx_list):
        img_t, gt_t = t2_ds[i]
        gt = gt_t.numpy()
        with torch.no_grad():
            logits = t2_net(img_t.unsqueeze(0).to(device))[1][0]      # Stage 2 logits (6, H, W)
        top2_idx = logits.topk(2, dim=0).indices.cpu().numpy()
        first, second = top2_idx[0], top2_idx[1]
        rescued_px = (first != gt) & (second == gt) & (gt > 0)
        img = img_t[0].numpy()
        panels = [np.dstack([img] * 3), t2_overlay(img, gt, 0.45), t2_overlay(img, first, 0.45), t2_overlay(img, second, 0.45),
                  t2_overlay(img, rescued_px.astype(int), 0.85, binary=True)]
        for c, p in enumerate(panels):
            axs[r, c].imshow(p, interpolation="nearest"); axs[r, c].axis("off")
            if r == 0:
                axs[r, c].set_title(heads[c], fontsize=10, fontweight="bold")
        axs[r, 0].text(0.01, 0.01, f"{filenames[i]} | {int(rescued[i]):,} px rescued", transform=axs[r, 0].transAxes,
                       fontsize=8, color="white", va="bottom", bbox=dict(facecolor="black", alpha=0.6, pad=2, edgecolor="none"))
    handles = [Patch(facecolor=T2_COLORS[n], edgecolor="black", label=n) for n in T2_LESIONS]
    handles.append(Patch(facecolor=np.array(T2_ORANGE) / 255.0, edgecolor="black", label="Rescued pixels"))
    fig.legend(handles=handles, loc="lower center", ncol=6, frameon=True, fontsize=9)
    plt.tight_layout(rect=[0, 0.05, 1, 1])
    t2_save(f"{T2_KEY}_{T2_SPLIT}_top2_rescued_examples", fig)
    plt.show()
else:
    print("No slices where the 2nd choice rescues lesion pixels.")

del t2_net
cleanup_memory()
```

```python
# ======================= CELL B: shared evaluation toolkit (IDENTICAL in every notebook) =======================
import os
import numpy as np
import pandas as pd
import torch
import matplotlib.pyplot as plt
from torch.utils.data import DataLoader
from tqdm.auto import tqdm
from matplotlib.patches import Patch

NUM_CLASSES  = 6
CLASS_NAMES  = ["Background", "SRF", "IRF", "PED", "SHRM", "IS/OS"]
LESION_NAMES = CLASS_NAMES[1:]
EVAL_PALETTE = np.array([[0, 0, 0], [255, 0, 0], [0, 255, 0], [0, 0, 255], [255, 255, 0], [255, 0, 255]], dtype=np.uint8)
EVAL_COLORS  = {n: "#%02x%02x%02x" % tuple(int(v) for v in rgb) for n, rgb in zip(CLASS_NAMES, EVAL_PALETTE)}
BINARY_COLOR = (255, 165, 0)                 # orange overlay for the binary (lesion vs background) masks
EVAL_SEED    = 42
OUT_DIR      = "eval_outputs"                # per-image scores / confusion matrices shared by all notebooks
os.makedirs(OUT_DIR, exist_ok=True)

if "device" not in globals():
    device = torch.device("mps" if torch.backends.mps.is_available() else "cuda" if torch.cuda.is_available() else "cpu")

EVAL_CONVENTIONS = {
    "exclude":    "class absent in BOTH ground truth and prediction -> slice dropped (NaN)",
    "credit":     "class absent in BOTH ground truth and prediction -> slice scores 100%",
    "gt_present": "only slices whose GROUND TRUTH contains the class are scored",
}
PRIMARY_CONVENTION = "exclude"


def ev_save(stem, fig=None):
    """Use save_fig (Cell A) when available, otherwise a plain PNG."""
    fig = fig if fig is not None else plt.gcf()
    if "save_fig" in globals():
        return save_fig(stem, fig=fig)
    fig.savefig(f"{stem}.png", dpi=200, bbox_inches="tight")


def ev_pick_split(preferred="test", data_root="processed_data"):
    if os.path.exists(os.path.join(data_root, preferred, "inputs")):
        return preferred
    print(f"WARNING: '{data_root}/{preferred}/inputs' not found -> using the 'val' split "
          f"(it was used for checkpoint selection, so numbers are slightly optimistic).")
    return "val"


# --------------------------------------------------------------------------------------------
# 1. One inference pass -> per-image confusion matrices (top-1, top-2) and binary 2x2 counts
# --------------------------------------------------------------------------------------------
@torch.no_grad()
def ev_collect(model, loader, mc_fn, bin_fn=None, desc=""):
    """
    cms1[i, gt, pred]  : 6x6 pixel confusion matrix, prediction = arg-max (top-1)
    cms2[i, gt, pred]  : same, but a pixel counts as correct if the ground truth is among the 2 most probable classes
                         (otherwise the top-1 label is kept as the prediction)
    bcms[i]            : [TN, FP, FN, TP] of the binary lesion head (sigmoid > 0.5), only if bin_fn is given
    """
    model.eval()
    cms1, cms2, bcms, means, stds = [], [], [], [], []
    for batch in tqdm(loader, desc=desc):
        images, targets = batch[0], batch[1]
        out = model(images.to(device))
        logits = mc_fn(out)
        top2 = logits.topk(2, dim=1).indices.cpu()                     # (B, 2, H, W)
        pred1 = top2[:, 0]
        hit2 = (top2 == targets.unsqueeze(1)).any(dim=1)
        pred2 = torch.where(hit2, targets, pred1)
        if bin_fn is not None:
            bpred = (torch.sigmoid(bin_fn(out)).squeeze(1) > 0.5).long().cpu()
        flat = images.flatten(1)
        means.append(flat.mean(dim=1).numpy())
        stds.append(flat.std(dim=1).numpy())
        for b in range(pred1.shape[0]):
            t = targets[b].reshape(-1)
            cms1.append(torch.bincount(t * NUM_CLASSES + pred1[b].reshape(-1), minlength=NUM_CLASSES ** 2)
                        .reshape(NUM_CLASSES, NUM_CLASSES).numpy())
            cms2.append(torch.bincount(t * NUM_CLASSES + pred2[b].reshape(-1), minlength=NUM_CLASSES ** 2)
                        .reshape(NUM_CLASSES, NUM_CLASSES).numpy())
            if bin_fn is not None:
                bcms.append(torch.bincount((t > 0).long() * 2 + bpred[b].reshape(-1), minlength=4).numpy())
    return (np.stack(cms1).astype(np.int64), np.stack(cms2).astype(np.int64),
            np.stack(bcms).astype(np.int64) if bin_fn is not None else None,
            np.concatenate(means), np.concatenate(stds))


def ev_get_bundle(key, build_fn, ckpt_path, make_dataset, mc_fn, bin_fn, split, batch_size=4, use_cache=True):
    """Runs the model once on `split` (cached in eval_outputs/ and invalidated when the checkpoint changes)."""
    cache = os.path.join(OUT_DIR, f"{key}_{split}_bundle.npz")
    mtime = os.path.getmtime(ckpt_path)
    if use_cache and os.path.exists(cache):
        z = np.load(cache)
        if float(z["ckpt_mtime"]) == mtime:
            print(f"[cache] loaded '{cache}'")
            return {"key": key, "split": split, "filenames": z["filenames"], "cms1": z["cms1"], "cms2": z["cms2"],
                    "bcms": z["bcms"] if bool(z["has_bin"]) else None, "img_mean": z["img_mean"], "img_std": z["img_std"]}
        print(f"[cache] '{cache}' is stale (checkpoint changed) -> recomputing")
    ds = make_dataset(split)
    loader = DataLoader(ds, batch_size=batch_size, shuffle=False, num_workers=0)
    net = build_fn().to(device)
    net.load_state_dict(torch.load(ckpt_path, map_location=device))
    cms1, cms2, bcms, mean, std = ev_collect(net, loader, mc_fn, bin_fn, desc=f"Evaluating {key} [{split}]")
    del net
    if torch.cuda.is_available():
        torch.cuda.empty_cache()
    elif torch.backends.mps.is_available():
        torch.mps.empty_cache()
    filenames = np.array(ds.filenames)
    np.savez_compressed(cache, cms1=cms1, cms2=cms2, bcms=(bcms if bcms is not None else np.zeros((0, 4), np.int64)),
                        has_bin=bcms is not None, filenames=filenames, img_mean=mean, img_std=std, ckpt_mtime=mtime)
    return {"key": key, "split": split, "filenames": filenames, "cms1": cms1, "cms2": cms2, "bcms": bcms,
            "img_mean": mean, "img_std": std}


# --------------------------------------------------------------------------------------------
# 2. Metrics from confusion matrices
# --------------------------------------------------------------------------------------------
def ev_class_stats(tot, c):
    inter = float(tot[c, c]); gt = float(tot[c, :].sum()); pr = float(tot[:, c].sum())
    return gt, pr, inter, 200.0 * inter / max(gt + pr, 1.0), 100.0 * inter / max(gt + pr - inter, 1.0)


def ev_pooled_table(cms):
    """Pixel-pooled Dice / IoU per lesion class (this is what the training logs and Table 3 comparison report)."""
    tot = cms.sum(axis=0).astype(np.float64)
    rows = []
    for c, name in enumerate(LESION_NAMES, start=1):
        gt, pr, inter, dice, iou = ev_class_stats(tot, c)
        rows.append({"Class": name, "GT px": int(gt), "Pred px": int(pr), "Dice (%)": dice, "IoU (%)": iou})
    df = pd.DataFrame(rows)
    df["GT share of lesion px (%)"] = 100.0 * df["GT px"] / df["GT px"].sum()
    return df


def ev_mean_variants(cms):
    """Three different 'mean' definitions - shows whether the headline mean depends on class size."""
    df = ev_pooled_table(cms)
    w = df["GT px"] / df["GT px"].sum()
    bg = ev_class_stats(cms.sum(axis=0).astype(np.float64), 0)
    return pd.DataFrame({
        "mDice (%)": [df["Dice (%)"].mean(), (w * df["Dice (%)"]).sum(), (df["Dice (%)"].sum() + bg[3]) / 6],
        "mIoU (%)":  [df["IoU (%)"].mean(),  (w * df["IoU (%)"]).sum(),  (df["IoU (%)"].sum() + bg[4]) / 6],
    }, index=["Unweighted mean, 5 lesion classes  (what the notebooks report)",
              "Weighted by GT class size (pixels), 5 lesion classes",
              "Unweighted mean incl. Background (6 classes)"])


def ev_per_image_metrics(cms, convention=PRIMARY_CONVENTION):
    """Per-image, per-class Dice / IoU (%) from per-image confusion matrices under a scoring convention."""
    assert convention in EVAL_CONVENTIONS, f"unknown convention '{convention}'"
    out = {}
    for c, name in enumerate(LESION_NAMES, start=1):
        inter = cms[:, c, c].astype(np.float64)
        t_sum = cms[:, c, :].sum(axis=1).astype(np.float64)
        p_sum = cms[:, :, c].sum(axis=1).astype(np.float64)
        union, denom = t_sum + p_sum - inter, t_sum + p_sum
        dice = np.where(denom > 0, 2.0 * inter / np.maximum(denom, 1.0), np.nan) * 100.0
        iou = np.where(union > 0, inter / np.maximum(union, 1.0), np.nan) * 100.0
        if convention == "credit":
            dice[denom == 0] = 100.0; iou[denom == 0] = 100.0
        elif convention == "gt_present":
            dice[t_sum == 0] = np.nan; iou[t_sum == 0] = np.nan
        out[f"{name}_DSC"] = dice
        out[f"{name}_IoU"] = iou
    df = pd.DataFrame(out)
    df["Mean_DSC"] = df[[f"{n}_DSC" for n in LESION_NAMES]].mean(axis=1, skipna=True)
    df["Mean_IoU"] = df[[f"{n}_IoU" for n in LESION_NAMES]].mean(axis=1, skipna=True)
    return df


def ev_bcms_from_cms(cms):
    """Collapse the 6-class confusion matrices into binary [TN, FP, FN, TP] (lesion = any class 1-5)."""
    return np.stack([cms[:, 0, 0], cms[:, 0, 1:].sum(axis=1), cms[:, 1:, 0].sum(axis=1), cms[:, 1:, 1:].sum(axis=(1, 2))],
                    axis=1).astype(np.int64)


def ev_binary_scores(bcms, prefix, convention=PRIMARY_CONVENTION):
    tn, fp, fn, tp = [bcms[:, i].astype(np.float64) for i in range(4)]
    denom, union = 2 * tp + fp + fn, tp + fp + fn
    dice = np.where(denom > 0, 2 * tp / np.maximum(denom, 1.0), np.nan) * 100.0
    iou = np.where(union > 0, tp / np.maximum(union, 1.0), np.nan) * 100.0
    if convention == "credit":
        dice[denom == 0] = 100.0; iou[denom == 0] = 100.0
    elif convention == "gt_present":
        dice[(tp + fn) == 0] = np.nan; iou[(tp + fn) == 0] = np.nan
    return pd.DataFrame({f"{prefix}_DSC": dice, f"{prefix}_IoU": iou})


def ev_pooled_binary(bcms):
    tn, fp, fn, tp = [float(bcms[:, i].sum()) for i in range(4)]
    return 200.0 * tp / max(2 * tp + fp + fn, 1.0), 100.0 * tp / max(tp + fp + fn, 1.0)


# --------------------------------------------------------------------------------------------
# 3. Bootstrap confidence intervals
# --------------------------------------------------------------------------------------------
def ev_bootstrap_ci(values, n_boot=2000, ci=0.99, seed=EVAL_SEED, chunk=250):
    v = np.asarray(values, dtype=np.float64); v = v[~np.isnan(v)]
    if v.size == 0:
        return np.nan, np.nan, np.nan, 0
    rng = np.random.default_rng(seed); means = np.empty(n_boot)
    for s in range(0, n_boot, chunk):
        e = min(n_boot, s + chunk)
        means[s:e] = v[rng.integers(0, v.size, size=(e - s, v.size))].mean(axis=1)
    a = (1.0 - ci) / 2.0 * 100.0
    return float(v.mean()), float(np.percentile(means, a)), float(np.percentile(means, 100.0 - a)), int(v.size)


def ev_ci_table(df_scores, columns, n_boot=2000, ci=0.99):
    rows = []
    for col in columns:
        m, lo, hi, n = ev_bootstrap_ci(df_scores[col].to_numpy(), n_boot, ci)
        rows.append({"Column": col, "N": n, "Mean": m, "CI lower": lo, "CI upper": hi})
    return pd.DataFrame(rows)


# --------------------------------------------------------------------------------------------
# 4. Confusion matrix (with Background) and overlay helpers for paper figures
# --------------------------------------------------------------------------------------------
def ev_plot_confusion(cm, title, stem):
    n6 = cm.astype(np.float64)
    row6 = n6.sum(axis=1, keepdims=True)
    norm6 = np.divide(n6, row6, out=np.zeros_like(n6), where=row6 > 0) * 100.0
    cm5 = n6[1:, 1:]; row5 = cm5.sum(axis=1, keepdims=True)
    norm5 = np.divide(cm5, row5, out=np.zeros_like(cm5), where=row5 > 0) * 100.0
    fig, axes = plt.subplots(1, 2, figsize=(16, 7))
    panels = [(axes[0], norm6, n6, CLASS_NAMES, "All 6 labels incl. Background (row-normalised %)"),
              (axes[1], norm5, cm5, LESION_NAMES, "Lesion classes only (GT lesion px predicted as a lesion)")]
    for ax, norm, counts, names, sub in panels:
        im = ax.imshow(norm, cmap="Blues", vmin=0, vmax=100)
        for i in range(len(names)):
            for j in range(len(names)):
                ax.text(j, i, f"{norm[i, j]:.1f}%\n({int(counts[i, j]):,})", ha="center", va="center", fontsize=8.5,
                        color="white" if norm[i, j] > 55 else "black")
        ax.set_xticks(range(len(names))); ax.set_yticks(range(len(names)))
        ax.set_xticklabels(names); ax.set_yticklabels(names)
        ax.set_xlabel("Predicted class"); ax.set_ylabel("Ground-truth class"); ax.set_title(sub, fontsize=11, fontweight="bold")
        fig.colorbar(im, ax=ax, fraction=0.046, pad=0.03, label="% of ground-truth pixels in row")
    fig.suptitle(title, fontsize=13, fontweight="bold")
    plt.tight_layout()
    ev_save(stem, fig)
    plt.show()
    return norm6, norm5


def ev_overlay(img, mask, alpha=0.45, binary=False):
    """Semi-transparent mask on top of the grayscale OCT slice. img: float [0,1] (H,W); mask: int labels (H,W)."""
    base = np.repeat((np.clip(img, 0, 1) * 255.0)[..., None], 3, axis=-1).astype(np.float32)
    m = np.asarray(mask) > 0
    color = np.zeros_like(base)
    if binary:
        color[:] = np.array(BINARY_COLOR, dtype=np.float32)
    else:
        color = EVAL_PALETTE[np.asarray(mask).astype(int)].astype(np.float32)
    out = base.copy()
    out[m] = (1.0 - alpha) * base[m] + alpha * color[m]
    return out.astype(np.uint8)


def ev_select_extremes(B, k=3, cls=None, min_gt_px=200, use="cms1"):
    """Indices of the k best and k worst slices by Dice (slices must contain the scored class(es) in the ground truth)."""
    df = ev_per_image_metrics(B[use], "gt_present")
    col = "Mean_DSC" if cls is None else f"{cls}_DSC"
    gt_px = B[use][:, 1:, :].sum(axis=(1, 2)) if cls is None else B[use][:, LESION_NAMES.index(cls) + 1, :].sum(axis=1)
    ok = df[col].notna().to_numpy() & (gt_px >= min_gt_px)
    sub = df.loc[ok, col]
    return sub.nlargest(k).index.tolist(), sub.nsmallest(k).index.tolist(), df


@torch.no_grad()
def ev_paper_figure(net, dataset, indices, mc_fn, bin_fn, notes, stem, alpha=0.45):
    """Rows = slices; columns = Input | GT multi-class overlay | Prediction multi-class overlay | GT binary | Prediction binary."""
    net.eval()
    n = len(indices)
    fig, axs = plt.subplots(n, 5, figsize=(16, 3.3 * n + 0.9), squeeze=False)
    heads = ["Input OCT", "Ground truth (multi-class)", "Prediction (multi-class)", "Ground truth (binary)",
             "Prediction (binary head)" if bin_fn is not None else "Prediction (binary, from multi-class)"]
    for r, (idx, note) in enumerate(zip(indices, notes)):
        sample = dataset[int(idx)]
        img_t, gt = sample[0], sample[1].numpy()
        out = net(img_t.unsqueeze(0).to(device))
        pred = mc_fn(out).argmax(dim=1)[0].cpu().numpy()
        bpred = ((torch.sigmoid(bin_fn(out))[0, 0] > 0.5).cpu().numpy().astype(int) if bin_fn is not None else (pred > 0).astype(int))
        img = img_t[0].numpy()
        panels = [np.dstack([img] * 3), ev_overlay(img, gt, alpha), ev_overlay(img, pred, alpha),
                  ev_overlay(img, (gt > 0).astype(int), alpha, binary=True), ev_overlay(img, bpred, alpha, binary=True)]
        for c, p in enumerate(panels):
            axs[r, c].imshow(p, cmap="gray" if c == 0 else None, interpolation="nearest"); axs[r, c].axis("off")
            if r == 0:
                axs[r, c].set_title(heads[c], fontsize=10, fontweight="bold")
        axs[r, 0].text(0.01, 0.01, note, transform=axs[r, 0].transAxes, fontsize=8, color="white", va="bottom",
                       bbox=dict(facecolor="black", alpha=0.6, pad=2, edgecolor="none"))
    handles = [Patch(facecolor=EVAL_COLORS[nm], edgecolor="black", label=nm) for nm in LESION_NAMES]
    handles.append(Patch(facecolor=np.array(BINARY_COLOR) / 255.0, edgecolor="black", label="Lesion (binary)"))
    fig.legend(handles=handles, loc="lower center", ncol=6, frameon=True, fontsize=9)
    plt.tight_layout(rect=[0, 0.05, 1, 1])
    ev_save(stem, fig)
    plt.show()
```

```python
# ===== CELL C (Model 2) =====
EVAL_KEY      = "model2_cascaded"
EVAL_LABEL    = "Model 2 Cascaded"
EVAL_CKPT     = "best_unet_model2.pth"
EVAL_BUILD    = lambda: CascadedUNetModel2(in_channels=1, num_classes=6, base_channels=64, concatenate_input=False)
EVAL_MAKE_DS  = lambda split: AMDDataset("processed_data", split=split, target_size=(512, 512), augment=False)
EVAL_MC_FN    = lambda out: out[1]          # model 2 returns (stage-1 binary, stage-2 multi-class)
EVAL_BIN_FN   = lambda out: out[0]
EVAL_BS       = 2 if device.type == "mps" else 4
```

```python
# ======================= CELL D: run the evaluation (identical in every notebook) =======================
# One pass over the evaluation split gives: top-1 and top-2 metrics, binary IoU/Dice, per-image scores, 99% bootstrap CIs.

def ev_report(B, label, n_boot=2000, ci=0.99, conv=PRIMARY_CONVENTION):
    key, split, c1, c2 = B["key"], B["split"], B["cms1"], B["cms2"]
    print("=" * 100)
    print(f"{label} | split='{split}' | {len(c1)} slices")
    print("=" * 100)

    # ---- 1. pooled multi-class metrics: top-1 vs top-2 -------------------------------------------------
    t1, t2 = ev_pooled_table(c1), ev_pooled_table(c2)
    comp = pd.DataFrame({"Class": t1["Class"], "GT share of lesion px (%)": t1["GT share of lesion px (%)"].round(2),
                         "Dice top-1 (%)": t1["Dice (%)"].round(2), "Dice top-2 (%)": t2["Dice (%)"].round(2),
                         "IoU top-1 (%)": t1["IoU (%)"].round(2), "IoU top-2 (%)": t2["IoU (%)"].round(2)})
    print("\nPixel-pooled per-class metrics (top-2: a pixel is correct if the GT class is one of the 2 most probable classes):")
    print(comp.to_string(index=False))

    # ---- 2. is the mean weighted by class size? ---------------------------------------------------------
    print("\nMean variants (top-1 predictions):")
    print(ev_mean_variants(c1).round(2).to_string())
    print("Mean variants (top-2 predictions):")
    print(ev_mean_variants(c2).round(2).to_string())

    # ---- 3. binary (lesion vs background) ---------------------------------------------------------------
    b_mc = ev_bcms_from_cms(c1)
    rows = [{"Binary source": "collapsed multi-class prediction (pred > 0)", "Dice (%)": ev_pooled_binary(b_mc)[0], "IoU (%)": ev_pooled_binary(b_mc)[1]}]
    if B["bcms"] is not None:
        rows.append({"Binary source": "binary head (sigmoid > 0.5)", "Dice (%)": ev_pooled_binary(B["bcms"])[0], "IoU (%)": ev_pooled_binary(B["bcms"])[1]})
    print("\nBinary lesion-vs-background metrics (pixel-pooled):")
    print(pd.DataFrame(rows).round(2).to_string(index=False))

    # ---- 4. per-image scores (+ binary) saved for the cross-model comparison (Cell G) ---------------------
    fn = pd.DataFrame({"filename": B["filenames"]})
    top1 = pd.concat([fn, ev_per_image_metrics(c1, conv), ev_binary_scores(b_mc, "BinMC", conv)], axis=1)
    if B["bcms"] is not None:
        top1 = pd.concat([top1, ev_binary_scores(B["bcms"], "BinHead", conv)], axis=1)
    top2 = pd.concat([fn, ev_per_image_metrics(c2, conv)], axis=1)
    top1.to_csv(os.path.join(OUT_DIR, f"{key}_{split}_scores_top1_{conv}.csv"), index=False)
    top2.to_csv(os.path.join(OUT_DIR, f"{key}_{split}_scores_top2_{conv}.csv"), index=False)
    pd.DataFrame(c1.sum(axis=0), index=[f"GT:{n}" for n in CLASS_NAMES], columns=[f"Pred:{n}" for n in CLASS_NAMES]) \
      .to_csv(os.path.join(OUT_DIR, f"{key}_{split}_confusion_counts_top1.csv"))

    # ---- 5. 99% bootstrap CIs (per-image means, convention='exclude') ------------------------------------------
    cols = [f"{n}_DSC" for n in LESION_NAMES] + ["Mean_DSC"] + [f"{n}_IoU" for n in LESION_NAMES] + ["Mean_IoU", "BinMC_DSC", "BinMC_IoU"]
    if B["bcms"] is not None:
        cols += ["BinHead_DSC", "BinHead_IoU"]
    ci_top1 = ev_ci_table(top1, cols, n_boot, ci).round(2)
    ci_top1.to_csv(os.path.join(OUT_DIR, f"{key}_{split}_ci{int(ci * 100)}_top1.csv"), index=False)
    print(f"\n{int(ci * 100)}% bootstrap CIs of the per-image mean (top-1, convention='{conv}'):")
    print(ci_top1.to_string(index=False))
    return top1, top2


EVAL_SPLIT = ev_pick_split("test")
B = ev_get_bundle(EVAL_KEY, EVAL_BUILD, EVAL_CKPT, EVAL_MAKE_DS, EVAL_MC_FN, EVAL_BIN_FN, EVAL_SPLIT, batch_size=EVAL_BS)
SCORES_TOP1, SCORES_TOP2 = ev_report(B, EVAL_LABEL)
```

```python
# ======================= CELL E: pixel confusion matrix including Background (identical in every notebook) =======================
cm_total = B["cms1"].sum(axis=0)
norm6, norm5 = ev_plot_confusion(cm_total, f"{EVAL_LABEL}: pixel confusion matrix, {B['split']} split ({len(B['cms1'])} slices)",
                                 stem=f"{EVAL_KEY}_{B['split']}_confusion_matrix")

conf = pd.DataFrame(norm6, index=CLASS_NAMES, columns=CLASS_NAMES)
pairs = [(g, p, conf.loc[g, p]) for g in CLASS_NAMES for p in CLASS_NAMES if g != p]
print("Top-8 confusions (share of the ground-truth class labelled as the other class):")
for g, p, v in sorted(pairs, key=lambda t: -t[2])[:8]:
    print(f"  GT {g:<10} -> predicted {p:<10}: {v:6.2f}%")
```

```python
# ======================= CELL J: top-2 predictions (final analysis; identical in every notebook) =======================
# Top-2 rule: a pixel counts as correct if the ground-truth class is one of the model's TWO most probable classes.
# This uses the ground truth, so it measures how close the model is, not a deployable result.
t1, t2 = ev_pooled_table(B["cms1"]), ev_pooled_table(B["cms2"])
tot1, tot2 = B["cms1"].sum(axis=0).astype(np.float64), B["cms2"].sum(axis=0).astype(np.float64)
cats = LESION_NAMES + ["Mean"]

d1, d2 = list(t1["Dice (%)"]) + [t1["Dice (%)"].mean()], list(t2["Dice (%)"]) + [t2["Dice (%)"].mean()]
i1, i2 = list(t1["IoU (%)"]) + [t1["IoU (%)"].mean()], list(t2["IoU (%)"]) + [t2["IoU (%)"].mean()]
recovered = []                                   # share of ground-truth pixels missed by top-1 that the 2nd choice gets right
for c in range(1, NUM_CLASSES):
    missed = tot1[c, :].sum() - tot1[c, c]
    recovered.append(100.0 * (tot2[c, c] - tot1[c, c]) / max(missed, 1.0))
recovered.append(float(np.mean(recovered)))

top2_df = pd.DataFrame({"Class": cats, "Dice top-1 (%)": d1, "Dice top-2 (%)": d2, "Dice gain (pts)": np.array(d2) - np.array(d1),
                        "IoU top-1 (%)": i1, "IoU top-2 (%)": i2, "IoU gain (pts)": np.array(i2) - np.array(i1),
                        "Missed px recovered by 2nd choice (%)": recovered}).round(2)
print(f"{EVAL_LABEL} | split='{B['split']}' | top-1 vs top-2 (pixel-pooled; Mean = unweighted mean of the 5 lesion classes)")
print(top2_df.to_string(index=False))
top2_df.to_csv(os.path.join(OUT_DIR, f"{EVAL_KEY}_{B['split']}_top1_vs_top2.csv"), index=False)

# ---- 99% bootstrap CIs of the per-image top-2 scores ----
ci_cols = [f"{n}_DSC" for n in LESION_NAMES] + ["Mean_DSC"] + [f"{n}_IoU" for n in LESION_NAMES] + ["Mean_IoU"]
ci_top2 = ev_ci_table(SCORES_TOP2, ci_cols).round(2)
ci_top2.to_csv(os.path.join(OUT_DIR, f"{EVAL_KEY}_{B['split']}_ci99_top2.csv"), index=False)
print("\n99% bootstrap CIs of the per-image mean, top-2 predictions:")
print(ci_top2.to_string(index=False))

# ---- bar chart: top-1 (solid) vs top-2 (hatched), bar colour = lesion class ----
fig, axes = plt.subplots(1, 2, figsize=(16, 5.5))
x, w = np.arange(len(cats)), 0.38
for ax, metric, a, b in zip(axes, ["Dice (DSC)", "IoU"], [d1, i1], [d2, i2]):
    cols = [PLOT_COLORS[c] for c in cats]
    ax.bar(x - w / 2, a, w, color=cols, edgecolor="black")
    r2 = ax.bar(x + w / 2, b, w, color=cols, edgecolor="black", hatch="//", alpha=0.8)
    for bar in r2:
        ax.annotate(f"{bar.get_height():.1f}", (bar.get_x() + bar.get_width() / 2, bar.get_height()), xytext=(0, 3),
                    textcoords="offset points", ha="center", va="bottom", fontsize=8, fontweight="bold")
    ax.set_xticks(x); ax.set_xticklabels(cats); ax.set_ylim(0, 105)
    ax.set_ylabel(f"{metric} (%)"); ax.grid(axis="y", linestyle="--", alpha=0.5)
    ax.set_title(f"{EVAL_LABEL}: {metric}, top-1 vs top-2 ({B['split']})", fontweight="bold")
axes[0].legend(handles=[Patch(facecolor="white", edgecolor="black", label="Top-1 (solid)"),
                        Patch(facecolor="white", edgecolor="black", hatch="//", label="Top-2 (hatched)")], loc="lower right")
plt.tight_layout()
ev_save(f"{EVAL_KEY}_{B['split']}_top1_vs_top2", fig)
plt.show()

# ---- slices where the 2nd choice rescues the most lesion pixels ----
diag = np.arange(1, NUM_CLASSES)
rescued = (B["cms2"][:, diag, diag] - B["cms1"][:, diag, diag]).sum(axis=1)
idx_list = [int(i) for i in np.argsort(-rescued)[:3] if rescued[i] > 0]
if idx_list:
    net = EVAL_BUILD().to(device)
    net.load_state_dict(torch.load(EVAL_CKPT, map_location=device))
    net.eval()
    ds = EVAL_MAKE_DS(B["split"])
    heads = ["Input OCT", "Ground truth", "Top-1 prediction", "2nd most probable class", "Rescued by 2nd choice (orange)"]
    fig, axs = plt.subplots(len(idx_list), 5, figsize=(16, 3.3 * len(idx_list) + 0.9), squeeze=False)
    for r, i in enumerate(idx_list):
        sample = ds[i]
        img_t, gt = sample[0], sample[1].numpy()
        with torch.no_grad():
            logits = EVAL_MC_FN(net(img_t.unsqueeze(0).to(device)))[0]
        top2_idx = logits.topk(2, dim=0).indices.cpu().numpy()
        first, second = top2_idx[0], top2_idx[1]
        rescued_px = (first != gt) & (second == gt) & (gt > 0)
        img = img_t[0].numpy()
        panels = [np.dstack([img] * 3), ev_overlay(img, gt, 0.45), ev_overlay(img, first, 0.45), ev_overlay(img, second, 0.45),
                  ev_overlay(img, rescued_px.astype(int), 0.85, binary=True)]
        for c, p in enumerate(panels):
            axs[r, c].imshow(p, interpolation="nearest"); axs[r, c].axis("off")
            if r == 0:
                axs[r, c].set_title(heads[c], fontsize=10, fontweight="bold")
        axs[r, 0].text(0.01, 0.01, f"{B['filenames'][i]} | {int(rescued[i]):,} px rescued", transform=axs[r, 0].transAxes,
                       fontsize=8, color="white", va="bottom", bbox=dict(facecolor="black", alpha=0.6, pad=2, edgecolor="none"))
    handles = seg_legend_handles(include_background=False) + [Patch(facecolor=np.array(BINARY_COLOR) / 255.0, edgecolor="black", label="Rescued pixels")]
    fig.legend(handles=handles, loc="lower center", ncol=6, frameon=True, fontsize=9)
    plt.tight_layout(rect=[0, 0.05, 1, 1])
    ev_save(f"{EVAL_KEY}_{B['split']}_top2_rescued_examples", fig)
    plt.show()
    del net
```

```python
# ===== CELL X: export the figure components (run in EVERY notebook, after Cell C + D, and after cell S) =====
# Saves img / GT / top-1 / 2nd-choice / binary masks for the chosen slices to eval_outputs/components_<key>_<split>.npz
# so figures can be redrawn later (cell F) without loading any model.
import json, cv2

SHARED_JSON = os.path.join(OUT_DIR, f"shared_slices_{B['split']}.json")
EXTRA_FILENAMES = []                         # optional: more slices, e.g. ["129_1_input.png"]
if os.path.exists(SHARED_JSON):
    sel = json.load(open(SHARED_JSON)); names = sel["high"] + sel["mid"] + sel["low"]
else:                                        # fallback until cell S has been run: best / median / worst slice of THIS model
    s = SCORES_TOP1.set_index("filename")["Mean_DSC"].dropna().sort_values()
    names = [s.index[-1], s.index[len(s) // 2], s.index[0]]
    print("shared_slices json not found -> exporting this model's own best / median / worst slice")
names = list(dict.fromkeys(names + EXTRA_FILENAMES))

def _rs(a, interp):                          # everything is stored at 512 x 512 so models with other input sizes line up
    return a if a.shape[:2] == (512, 512) else cv2.resize(a, (512, 512), interpolation=interp)

@torch.no_grad()
def export_components(names):
    ds = EVAL_MAKE_DS(B["split"]); pos = {n: i for i, n in enumerate(ds.filenames)}
    net = EVAL_BUILD().to(device); net.load_state_dict(torch.load(EVAL_CKPT, map_location=device)); net.eval()
    keep = {k: [] for k in ["img", "gt", "pred1", "pred2", "bin_pred", "bin_prob"]}
    for n in names:
        sample = ds[pos[n]]
        img_t, gt = sample[0], sample[1].numpy()
        out = net(img_t.unsqueeze(0).to(device))
        top2 = EVAL_MC_FN(out)[0].topk(2, dim=0).indices.cpu().numpy()
        if EVAL_BIN_FN is not None:
            prob = torch.sigmoid(EVAL_BIN_FN(out))[0, 0].cpu().numpy(); bpred = (prob > 0.5)
        else:
            prob = (top2[0] > 0).astype(np.float32); bpred = top2[0] > 0
        keep["img"].append(_rs((img_t[0].numpy() * 255).round().astype(np.uint8), cv2.INTER_LINEAR))
        keep["gt"].append(_rs(gt.astype(np.uint8), cv2.INTER_NEAREST))
        keep["pred1"].append(_rs(top2[0].astype(np.uint8), cv2.INTER_NEAREST))
        keep["pred2"].append(_rs(top2[1].astype(np.uint8), cv2.INTER_NEAREST))
        keep["bin_pred"].append(_rs(bpred.astype(np.uint8), cv2.INTER_NEAREST))
        keep["bin_prob"].append(_rs((prob * 255).round().astype(np.uint8), cv2.INTER_LINEAR))
    path = os.path.join(OUT_DIR, f"components_{EVAL_KEY}_{B['split']}.npz")
    np.savez_compressed(path, filenames=np.array(names), has_bin=EVAL_BIN_FN is not None,
                        pixel_threshold=int(globals().get("PIXEL_THRESHOLD", 0)), label=EVAL_LABEL,
                        **{k: np.stack(v) for k, v in keep.items()})
    print(f"saved {len(names)} slices -> {path}")
    del net

export_components(names)
```

```python
# ===== CELL B2: cleaning functions (+ one picture to check them) =====
import glob, cv2
from scipy import ndimage as ndi

def rgb_to_class_clean(mask_rgb, tol=60):
    """Like the notebooks' _rgb_to_class, but blended edge pixels (> tol from every class colour) take the label of the
    NEAREST clean pixel instead of the 'closest class colour' (which can be IRF/green)."""
    m = np.ascontiguousarray(mask_rgb, dtype=np.int32)
    d2 = ((m[:, :, None, :] - EVAL_PALETTE.astype(np.int32)[None, None]) ** 2).sum(-1)
    lab = d2.argmin(-1)
    bad = d2.min(-1) > tol ** 2
    if bad.any() and (~bad).any():
        idx = ndi.distance_transform_edt(bad, return_distances=False, return_indices=True)
        lab = lab[idx[0], idx[1]]
    return lab

def remove_thin_irf(lab, irf=2, k=3):
    """Relabel thin IRF 'outline' pixels (touching another lesion class) with the nearest non-outline label. Works on GT or predictions."""
    irf_m = lab == irf
    core = ndi.binary_opening(irf_m, structure=np.ones((k, k), bool))
    other = (lab > 0) & ~irf_m
    near_other = ndi.binary_dilation(other, structure=np.ones((3, 3), bool), iterations=2)
    thin = irf_m & ~core & near_other
    if not thin.any() or thin.all():
        return lab
    idx = ndi.distance_transform_edt(thin, return_distances=False, return_indices=True)
    return lab[idx[0], idx[1]]

# ---- picture: raw GT colours | current labels | cleaned labels | pixels that changed ----
IMG_NO = 0                                   # change to look at other masks
f = sorted(glob.glob("processed_data/val/outputs/*.png"))[IMG_NO]
rgb = cv2.cvtColor(cv2.imread(f), cv2.COLOR_BGR2RGB)
old = ((rgb.astype(np.int32)[:, :, None, :] - EVAL_PALETTE.astype(np.int32)[None, None]) ** 2).sum(-1).argmin(-1)
new = remove_thin_irf(rgb_to_class_clean(rgb))
fig, axs = plt.subplots(1, 4, figsize=(16, 4.5))
for ax, im, t in zip(axs, [rgb, EVAL_PALETTE[old], EVAL_PALETTE[new], (old != new)], ["Raw GT colours", "Current labels (argmin)", "Cleaned labels", "Pixels changed"]):
    ax.imshow(im, cmap="gray", interpolation="nearest"); ax.set_title(t, fontweight="bold"); ax.axis("off")
plt.tight_layout(); ev_save("irf_cleanup_check", fig); plt.show()
print(f"{os.path.basename(f)}: IRF px  current={int((old == 2).sum())}  cleaned={int((new == 2).sum())}  | changed px={int((old != new).sum())}")
```

```python
# ===== CELL B3: re-score the EXISTING model with cleaned GT / cleaned predictions (no retraining) =====
# Run in a model notebook after Cell C + Cell D + Cell B2. Shows whether IRF Dice changes when the outline is removed.
@torch.no_grad()
def irf_cleanup_check(max_slices=None, bs=2):
    ds, ds_c = EVAL_MAKE_DS(B["split"]), EVAL_MAKE_DS(B["split"])
    ds_c._rgb_to_class = rgb_to_class_clean               # same dataset, but cleaned label conversion
    net = EVAL_BUILD().to(device); net.load_state_dict(torch.load(EVAL_CKPT, map_location=device)); net.eval()
    n = len(ds) if max_slices is None else min(max_slices, len(ds))
    names = ["original GT / original pred", "original GT / cleaned pred", "cleaned GT / cleaned pred"]
    cms = {k: np.zeros((6, 6), np.int64) for k in names}
    for s in tqdm(range(0, n, bs), desc="IRF clean-up check"):
        ids = list(range(s, min(s + bs, n)))
        items = [ds[i] for i in ids]
        imgs = torch.stack([it[0] for it in items]).to(device)
        gt_o = torch.stack([it[1] for it in items]).numpy()
        gt_c = torch.stack([ds_c[i][1] for i in ids]).numpy()
        p = EVAL_MC_FN(net(imgs)).argmax(1).cpu().numpy()
        pc = np.stack([remove_thin_irf(x) for x in p])
        for k, (g, q) in zip(names, [(gt_o, p), (gt_o, pc), (gt_c, pc)]):
            cms[k] += np.bincount((g * 6 + q).ravel(), minlength=36).reshape(6, 6)
    out = pd.DataFrame({"Class": LESION_NAMES + ["Mean"]})
    for k in names:
        t = ev_pooled_table(cms[k][None])
        out[f"Dice | {k}"] = list(t["Dice (%)"]) + [t["Dice (%)"].mean()]
    print(f"{EVAL_LABEL} | split={B['split']} | pooled Dice (%)")
    print(out.round(2).to_string(index=False))
    out.round(2).to_csv(os.path.join(OUT_DIR, f"{EVAL_KEY}_{B['split']}_irf_cleanup_dice.csv"), index=False)

irf_cleanup_check()          # tip: irf_cleanup_check(max_slices=100) for a quick test
```
